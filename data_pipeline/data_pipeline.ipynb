#Automating url injection and scraping data from different urls.

```python
master=[]
tag="https://books.toscrape.com/"
for category,last in categories.items():
    url=tag+last
    print(url)
    response=requests.get(url)
    soup=BeautifulSoup(response.text)
    articles=soup.find_all("article", class_="product_pod")
    for article in articles:
        title=article.select_one("h3 > a[title]").get("title")
        price=article.find("p",class_="price_color").text
        rating=article.find("p",class_="star-rating")["class"][1]
        availability=article.find("p",class_="instock availability").text.strip()
        record=[title,price,rating,availability,category]
        master.append(record)
```

#SQL-Table creation

```python
import sqlite3
```

```python
conn=sqlite3.connect("my_database.db")
Cursor=conn.cursor()
```

```python
Cursor.execute('''
CREATE TABLE categories(
category_id INTEGER PRIMARY KEY,
category_name TEXT UNIQUE);
''')
conn.commit()
```

```python
Cursor.execute('''
CREATE TABLE books(
book_id INTEGER PRIMARY KEY,
 title TEXT,
  price_gbp REAL,
   price_inr REAL,
    rating INTEGER,
     in_stock INTEGER,
      category_id INTEGER REFERENCES categories(category_id));
''')
conn.commit()
```

#Insertion completed and starting queries

```python
#select all books whose category id is 1,5,7
query1=pd.read_sql("SELECT * FROM books WHERE category_id IN (1,5,7)",conn)   #1st query
```

```python
# select title and price in inr of books whose category id is 1 with hishest prices
query2=pd.read_sql("SELECT title,price_inr FROM books WHERE category_id = 1 ORDER BY price_inr DESC",conn)   #2nd query
query2
```

```python
#select title and price in inr of 10 lowest priced books
query3=pd.read_sql("SELECT title,price_inr FROM books ORDER BY price_inr LIMIT 10",conn)   #3rd query
```

```python
#select all the books where the price in gbp is in between 20 and 40
query4=pd.read_sql("SELECT * from books WHERE price_gbp BETWEEN 20 AND 40",conn)       #4th query
```

```python
#select 20 unique category names from books on ascending order of category names
query5=pd.read_sql(
    """
    SELECT DISTINCT c.category_name
    FROM books AS b
    JOIN categories AS c
        ON b.category_id = c.category_id
    ORDER BY c.category_name
    LIMIT 20
    """,
    conn
)   #5th query
```

```python
#select 20 hishest prices books title,category name,prices in gbp,int and rating
query6=pd.read_sql(
    """
    SELECT
        b.title,
        c.category_name,
        b.price_gbp,
        b.price_inr,
        b.rating
    FROM books AS b
    INNER JOIN categories AS c
        ON b.category_id = c.category_id
    ORDER BY b.price_inr DESC
    LIMIT 20
    """,
    conn
)      #6th query
query6
```

```python
categories_df = pd.read_sql(
    "SELECT category_id, category_name FROM categories",
    conn
)

merge_result = pd.merge(
    df,
    categories_df,
    left_on="category",
    right_on="category_name",
    how="inner"
)

merge_result = merge_result[
    ["title", "category_name", "price_gbp", "price_inr", "rating"]
].sort_values(
    "price_inr",
    ascending=False
).head(20).reset_index(drop=True)
merge_result
```
